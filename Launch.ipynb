```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from huggingface_hub import login
from Streaming_prompt_simulation import chunk_text_by_words

your_token = 123 #
login(token=your_token)

model_name = "GSAI-ML/LLaDA-8B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True, token=your_token)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16,        
    trust_remote_code=True,
    token=your_token,
    device_map={"": "cuda:0"}          
)
model.eval()
```

```python
from RestartDiffusionEvaluator import RestartDiffusionEvaluator

evaluator = RestartDiffusionEvaluator(model, tokenizer)

system_prompt = "User: Answer the question. Assistant: "
text = "Which statement best explains why a leather car seat can feel hot on a sunny day?A) Thermal energy is destroyed by the seat.B) Thermal energy is absorbed by the seat.C) Kinetic energy is transformed into thermal energy.D) Chemical energy is transformed into thermal energy."
chunks = chunk_text_by_words(text=text)

current_prompt_ids = tokenizer.encode(system_prompt, add_special_tokens=False)

for idx, chunk in enumerate(chunks):
    chunk_ids = tokenizer.encode(chunk, add_special_tokens=False)
    current_prompt_ids.extend(chunk_ids)
    
    out_restart = evaluator.process_chunk(current_prompt_ids, steps_budget=6)

final_output = evaluator.finalize(current_prompt_ids, cleanup_steps=16)

print(f"\nОТВЕТ: {final_output}")
```

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM
from huggingface_hub import login
from Streaming_prompt_simulation import chunk_text_by_words

your_token = 123 #
login(token=your_token)

model_name = "Qwen/Qwen2.5-7B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(model_name, trust_remote_code=True, token=your_token)

model = AutoModelForCausalLM.from_pretrained(
    model_name,
    torch_dtype=torch.bfloat16,        
    trust_remote_code=True,
    token=your_token,
    device_map={"": "cuda:0"}          
)
model.eval()
```

```python
from StandardARModel import StandardARModel

ar_model = StandardARModel(model, tokenizer, device="cuda", base_tokens=32)

ar_model.set_system_prompt("You are a helpful assistant. Answer concisely.\n\n")

text = "Which statement best explains why a leather car seat can feel hot on a sunny day?A) Thermal energy is destroyed by the seat.B) Thermal energy is absorbed by the seat.C) Kinetic energy is transformed into thermal energy.D) Chemical energy is transformed into thermal energy."

chunks = chunk_text_by_words(text=text)

for chunk in chunks:
    ar_model.process_speech_chunk(chunk, compute_budget=0)

response = ar_model.finalize(additional_steps=64)
print(f"Ответ: {response}")
```
